```python
# Function to evaluate models


from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_absolute_error
import numpy as np

# Handle missing values
imputer = SimpleImputer(strategy="median")

x_train1 = imputer.fit_transform(x_train1)
x_val = imputer.transform(x_val)

# Train and evaluate models
def classifier(model, x_train1, y_train, x_val, y_val):
    model.fit(x_train1, y_train)
    y_pred = model.predict(x_val)

    r2 = r2_score(y_val, y_pred)
    mae = mean_absolute_error(y_val, y_pred)

    return r2, mae

for name, model in class_dicct.items():
    try:
        r2, mae = classifier(
            model, x_train1, y_train, x_val, y_val
        )
        print(f"{name}: R2 Score = {r2:.2f}")
        print(f"{name}: Mean Absolute Error = {mae:.2f}")
        print("-" * 40)

    except Exception as e:
        print(f"{name} failed: {e}")

```
